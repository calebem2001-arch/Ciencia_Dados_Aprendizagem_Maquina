## 1. Criando o DataFrame

## 27. DataFrame de salários

```python
cidades_salarios = pd.DataFrame([
    ['Paraná', 'Londrina', 1356.00],
    ['São Paulo', 'São Carlos', 1508.00],
    ['Santa Catarina', 'Florianópolis', 1798.00],
    ['Paraná', 'Curitiba', 2293.00],
    ['São Paulo', 'Campinas', 1710.00]
], columns=['Estado', 'Cidade', 'Salário-Médio'])

cidades_salarios
```

## 28. DataFrame de população

```python
cidades_pop = pd.DataFrame([
    ['Londrina', 575377],
    ['São Carlos', 254484],
    ['Florianópolis', 508826],
    ['Curitiba', 1963726],
    ['Campinas', 1223237]
], columns=['Cidade', 'Habitantes'])

cidades_pop
```

## 29. Juntando DataFrames com `merge()`

```python
resultado_merge = pd.merge(
    left=cidades_salarios,
    right=cidades_pop,
    on='Cidade'
)

resultado_merge
```

## 30. `merge()` usando `outer`

```python
resultado_outer = pd.merge(
    left=cidades_salarios,
    right=cidades_pop,
    on='Cidade',
    how='outer'
)

resultado_outer
```

## 31. Concatenando DataFrames com `concat()`

```python
cidades_df1 = pd.DataFrame([
    ['Londrina', 575377],
    ['São Carlos', 254484],
    ['Florianópolis', 508826]
], columns=['Cidade', 'Habitantes'])

cidades_df2 = pd.DataFrame([
    ['Curitiba', 1963726],
    ['Campinas', 1223237]
], columns=['Cidade', 'Habitantes'])

resultado_concat = pd.concat(
    [cidades_df1, cidades_df2],
    ignore_index=True
)

resultado_concat
```

. Salvando os dados em CSV, HTML e JSON

```python
cidades.to_csv('cidades.csv', index=False)
cidades.to_html('cidades.html', index=False)
cidades.to_json('cidades.json', orient='records', force_ascii=False)

print('Arquivos criados com sucesso:')
print('- cidades.csv')
print('- cidades.html')
print('- cidades.json')
```
