## 3. Matriz de confusão do exemplo do livro (Figura 12.1)
Classe 1 = positiva, classe 2 = negativa.

```python
VP, FN = 25, 0
FP, VN = 10, 40

TFP = FP / (FP + VN)                           # Eq. 12.5
TFN = FN / (FN + VP)                           # Eq. 12.6
TVP = VP / (FN + VP)                           # Eq. 12.7 (revocação)
precisao = VP / (FP + VP)                      # Eq. 12.8
especificidade = VN / (FP + VN)                # Eq. 12.9
acuracia = (VN + VP) / (VN + VP + FN + FP)     # Eq. 12.10
f1 = (2 * precisao * TVP) / (precisao + TVP)   # Eq. 12.12

print(f"TFP: {TFP:.3f}")
print(f"TFN: {TFN:.3f}")
print(f"TVP (Revocação): {TVP:.3f}")
print(f"Precisão: {precisao:.3f}")
print(f"Especificidade: {especificidade:.3f}")
print(f"Acurácia: {acuracia:.3f}")
print(f"F1-Score: {f1:.3f}")
```

## 5. Classificação

```python
import os
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (accuracy_score, f1_score, recall_score, precision_score,
                             roc_auc_score, classification_report,
                             confusion_matrix, ConfusionMatrixDisplay)

if os.path.exists('diabetes.csv'):
    df = pd.read_csv('diabetes.csv')
else:
    from sklearn.datasets import load_breast_cancer
    print("diabetes.csv não encontrado: usando dataset de exemplo (breast cancer).")
    bc = load_breast_cancer(as_frame=True)
    df = bc.frame  # a última coluna é o alvo

X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values

training_set, test_set, train_labels, test_labels = train_test_split(
    X, y, test_size=0.3, random_state=12, stratify=y)

dt = DecisionTreeClassifier(random_state=42)
dt.fit(training_set, train_labels)
preds_dt = dt.predict(test_set)

print(f"Acurácia: {accuracy_score(test_labels, preds_dt)}")
print(f"F1-Score: {f1_score(test_labels, preds_dt)}")
print(f"Revocação (Recall): {recall_score(test_labels, preds_dt)}")
print(f"Precisão: {precision_score(test_labels, preds_dt)}")
print(f"AUC: {roc_auc_score(test_labels, preds_dt)}")
print()
print(classification_report(test_labels, preds_dt))

ConfusionMatrixDisplay(confusion_matrix(test_labels, preds_dt)).plot()
plt.title("Matriz de Confusão")
plt.show()
```

## 6. Curva ROC e curva Precisão-Revocação

```python
from sklearn.metrics import roc_curve, precision_recall_curve, auc

probs = dt.predict_proba(test_set)[:, 1]

fpr, tpr, _ = roc_curve(test_labels, probs)
plt.figure()
plt.plot(fpr, tpr, marker='o', label=f"AUC = {auc(fpr, tpr):.3f}")
plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel("TFP"); plt.ylabel("TVP (revocação)")
plt.title("Curva ROC"); plt.legend(); plt.show()

prec, rec, _ = precision_recall_curve(test_labels, probs)
plt.figure()
plt.plot(rec, prec, marker='o', label=f"AUPRC = {auc(rec, prec):.3f}")
plt.xlabel("Revocação"); plt.ylabel("Precisão")
plt.title("Curva Precisão-Revocação"); plt.legend(); plt.show()
```

## 7. Modelo descritivo: índice silhueta

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler

X_scaled = StandardScaler().fit_transform(X)

for k in range(2, 7):
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    print(f"k={k} -> silhueta média: {silhouette_score(X_scaled, labels):.4f}")
```

## 8. Ajuste de hiperparâmetros: Grid Search e Random Search

```python
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
from scipy.stats import randint

param_grid = {
    'max_depth': [3, 5, 7, 10, None],
    'min_samples_split': [2, 5, 10],
    'criterion': ['gini', 'entropy']
}
gs = GridSearchCV(DecisionTreeClassifier(random_state=42), param_grid, cv=5, scoring='f1')
gs.fit(training_set, train_labels)
print("Grid Search - melhores parâmetros:", gs.best_params_)
print("Grid Search - melhor F1 (CV):", gs.best_score_)

param_dist = {
    'max_depth': randint(2, 15),
    'min_samples_split': randint(2, 20),
    'criterion': ['gini', 'entropy']
}
rs = RandomizedSearchCV(DecisionTreeClassifier(random_state=42), param_dist,
                        n_iter=20, cv=5, scoring='f1', random_state=42)
rs.fit(training_set, train_labels)
print("Random Search - melhores parâmetros:", rs.best_params_)
print("Random Search - melhor F1 (CV):", rs.best_score_)

melhor = gs.best_estimator_
print("F1 no teste:", f1_score(test_labels, melhor.predict(test_set)))
```

## 9. Teste de hipóteses: comparando dois algoritmos

```python
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.ensemble import RandomForestClassifier
from scipy import stats

cv = StratifiedKFold(n_splits=10, shuffle=True, random_state=12)

scores_dt = cross_val_score(DecisionTreeClassifier(random_state=42), X, y, cv=cv)
scores_rf = cross_val_score(RandomForestClassifier(random_state=42), X, y, cv=cv)

print(f"Árvore de Decisão: média = {scores_dt.mean():.4f}")
print(f"Random Forest:     média = {scores_rf.mean():.4f}")

print("Normalidade DT (Shapiro p):", stats.shapiro(scores_dt).pvalue)
print("Normalidade RF (Shapiro p):", stats.shapiro(scores_rf).pvalue)

t_stat, p_t = stats.ttest_rel(scores_dt, scores_rf)
w_stat, p_w = stats.wilcoxon(scores_dt, scores_rf)
print(f"t-test pareado: p = {p_t:.4f}")
print(f"Wilcoxon:       p = {p_w:.4f}")

alpha = 0.05  # 95% de confiança
if p_w < alpha:
    print("Rejeita H0: diferença estatisticamente significativa.")
else:
    print("Não rejeita H0: sem diferença estatisticamente significativa.")
```
